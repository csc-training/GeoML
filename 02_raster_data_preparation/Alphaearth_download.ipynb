Manual steps before:
* Read: https://source.coop/tge-labs/aef#manifest-and-index

* Select in QGIS relevant polygons

* Use `AWS_NO_SIGN_REQUEST=YES` in front of all GDAL commands, that downloads from S3 would work.
* There are both Https and S3 links. The .vrt files include S3 links to .tiff files.
* `--debug ON` is just to see that something is happening and to show errors. Can be removed.

* Download and save in 3067 with gdalwarp

## Get shallow learning bbox

```python
! AWS_NO_SIGN_REQUEST=YES gdalwarp --debug ON -r nearest \
  -t_srs EPSG:3067 \
  -te 270000 6690000 290000 6710000 \
  -tr 10 10 \
  /vsicurl/https://data.source.coop/tge-labs/aef/v1/annual/2021/34N/xg6o3outxe18szaej-0000008192-0000008192.vrt \
  alphaearth_shallow_3067.tiff
```
